```python
# Exercise 1: Regression - Optimize
"""
Exercise: Setting up optimization for a regression neural network

Neural Network Architecture:
- Input Layer: 5 features
- Hidden Layer 1: 8 neurons with sigmoid activation
- Hidden Layer 2: 4 neurons with sigmoid activation
- Output Layer: 1 neuron with LINEAR activation (key for regression!)

The linear activation in the output layer allows the network to predict
continuous values (any real number), which is essential for regression tasks.
"""

from tensorflow import keras
from keras.layers import Dense

# Build the regression neural network
model = keras.Sequential()
model.add(Dense(8, input_shape=(5,), activation='sigmoid'))
model.add(Dense(4, activation='sigmoid'))
model.add(Dense(1, activation='linear'))  # Linear activation for regression!


# SOLUTION 1: Using Mean Squared Error (MSE) - Most Common for Regression
model.compile(
    optimizer='adam',
    loss='mse',           # TODO1: Mean Squared Error
    metrics=['mae']       # TODO2: Mean Absolute Error
)

print("✓ Model compiled successfully with MSE loss and MAE metric!")
print("\nModel Summary:")
model.summary()

# Alternative 2: Using MAE as loss (good when you have outliers)
print("\n2. Using MAE as loss (more robust to outliers):")
print("""
model.compile(
    optimizer='adam',
    loss='mae',              # Mean Absolute Error as loss
    metrics=['mse']          # MSE as evaluation metric
)
""")
```

```python
# Exercise 2: Regression example
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam

columns = ["mpg", "cylinders", "displacement", "horsepower", "weight",
           "acceleration", "model year", "origin", "car name"]

df = pd.read_csv("auto-mpg.csv", na_values="?", comment="\t",
                 sep=",", skipinitialspace=True)
# Drop missing values
df = df.dropna()
print(df.head())
print(df.shape)
print(df.dtypes)
# Drop non-needed columns
df = df.drop(columns=["model year", "origin", "car name"])

# Features and target
X = df.drop(columns=["mpg"])
y = df["mpg"]

# Train/test split WITHOUT shuffling
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, shuffle=False
)

# Scale features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 3. BUILD THE MODEL (2 hidden layers)
model = Sequential([
    Dense(64, activation='relu', input_shape=(X_train_scaled.shape[1],)),
    Dense(64, activation='relu'),
    Dense(1)   # regression output
])

model.compile(
    optimizer=Adam(0.001),
    loss='mean_squared_error',
    metrics=['mean_absolute_error']
)

# 4. TRAIN
history = model.fit(
    X_train_scaled, y_train,
    epochs=200,   # increase if MAE is too high
    batch_size=16,
    verbose=1
)

# 5. EVALUATE ON SCALED TEST SET
loss, mae = model.evaluate(X_test_scaled, y_test, verbose=0)

print(f"Test MAE: {mae:.3f}")
```
